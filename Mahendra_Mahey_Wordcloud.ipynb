
## A Word Cloud Visualisation

This notebook explores descriptive adjectives associated
with Mahendra Mahey through an engaging word cloud.

### Objectives
- Explore descriptive language.
- Highlight creativity and innovation.
- Visualise professional qualities.
- Generate a high-resolution poster.

The word sizes represent their assigned importance,
not measured personality ratings.



## 1. Importing libraries

We use three main libraries:

- WordCloud: generates the word cloud.
- Matplotlib: displays the visualisation.
- Counter: calculates word frequencies.


```python

%pip install wordcloud matplotlib pillow
```

```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud
from collections import Counter

print("Libraries imported successfully!")
```


## 2. Creating the adjective dataset

The following adjectives describe five broad themes:

1. Personality and character
2. Professional qualities
3. Creativity and innovation
4. Leadership and collaboration
5. Intellectual and cultural qualities

These are interpretative descriptions rather than
independently verified personality assessments.


```python

words = ['Inspiring', 'Kind', 'Approachable', 'Generous', 'Enthusiastic', 'Passionate', 'Humble', 'Warm', 'Friendly', 'Empathetic', 'Thoughtful', 'Genuine', 'Respectful', 'Encouraging', 'Supportive', 'Open-minded', 'Curious', 'Optimistic', 'Patient', 'Compassionate', 'Innovative', 'Visionary', 'Strategic', 'Collaborative', 'Resourceful', 'Knowledgeable', 'Experienced', 'Adaptable', 'Proactive', 'Dedicated', 'Committed', 'Organised', 'Analytical', 'Reflective', 'Professional', 'Reliable', 'Responsible', 'Independent', 'Resilient', 'Multidisciplinary', 'Creative', 'Imaginative', 'Experimental', 'Inventive', 'Entrepreneurial', 'Exploratory', 'Original', 'Forward-thinking', 'Transformative', 'Ambitious', 'Adventurous', 'Progressive', 'Dynamic', 'Pioneering', 'Community-minded', 'Empowering', 'Inclusive', 'Democratic', 'Participatory', 'Trustworthy', 'Facilitative', 'Cooperative', 'Interdisciplinary', 'People-oriented', 'Diplomatic', 'Motivational', 'Enabling', 'Unifying', 'Accessible', 'Engaging', 'Appreciative', 'Inquisitive', 'Informed', 'Scholarly', 'Culturally aware', 'Digitally minded', 'Interconnected', 'Thought-provoking', 'Insightful', 'Open', 'Knowledge-sharing', 'Intercultural', 'Human-centred', 'Intergenerational', 'Impactful']

print("Number of adjectives:", len(words))
```

## We can also manualy include additional adjectives

```python
words.append("Fun")
words.append("Joy")
```


## 3. Assigning importance to selected adjectives

Word clouds use frequency values to determine
the relative size of words.

We increase the weights of selected adjectives
to make them visually prominent.


```python

frequencies = Counter(words)

important_words = {
    "Collaborative": 40,
    "Innovative": 38,
    "Inspiring": 36,
    "Fun": 35,
    "Visionary": 34,
    "Creative": 32,
    "Generous": 30,
    "Inclusive": 28,
    "Joy": 27,
    "Empowering": 27,
    "Experimental": 26,
    "Community-minded": 25,
    "Passionate": 24,
    "Transformative": 23
}

frequencies.update(important_words)

print("Word frequencies created successfully!")
```

```python

wc = WordCloud(
    width=2400,
    height=1600,
    background_color="#F7F7F2",
    max_words=100,
    prefer_horizontal=0.9,
    relative_scaling=0.6,
    margin=12,
    collocations=False
).generate_from_frequencies(frequencies)

print("Word cloud generated successfully!")
```


## 4. Displaying the word cloud

We use Matplotlib to display the final visualisation.

The largest words represent the adjectives
that have been assigned the highest weights.


```python

plt.figure(figsize=(18, 12))

plt.imshow(wc, interpolation="bilinear")
plt.axis("off")

plt.title(
    "MAHENDRA MAHEY",
    fontsize=36,
    fontweight="bold",
    color="#12355B",
    pad=30
)

plt.text(
    0.5, 0.02,
    "A celebration of creativity, collaboration and innovation",
    transform=plt.gca().transAxes,
    ha="center",
    fontsize=17,
    color="#087E8B"
)

plt.tight_layout()
plt.show()
```


## 6. Exporting the visualisation

We save the word cloud as a high-resolution PNG image.

This image can be used in:
- Presentations
- Posters
- Reports
- Workshops
- Digital exhibitions


```python

wc.to_file("mahendra_mahey_wordcloud.png")

print("Image saved as mahendra_mahey_wordcloud.png")
```
